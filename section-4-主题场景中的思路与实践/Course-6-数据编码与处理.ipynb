# 数据编码预处理
本节讲解Python处理常见编码形式的方法，比如csv,json,xml

```python
# 读写csv
import csv
with open('data.csv', 'r') as f:
    reader = csv.reader(f)
    headers = next(reader)
    for row in reader:
        print(row)
        
with open('data.csv', 'w') as f:
    writer = csv.writer(f)
    writer.writerow(headers)
    writer.writerows([
        ['a', 'b', 'c'] 
    ])
```

```python

import json
# 读取json
with open('data.json', 'r') as f:
    data = json.load(f)
    print(data)
    
# 写入json
data = {'name': 'John', 'age': 30}
with open('data.json', 'w') as f:
    json.dump(data, f)
    
```

```python
# 解析xml文档

from xml.dom.minidom import parse
xmlobj = parse('test.xml')

for node in xmlobj.getElementsByTagName('book'):
    print(node.getAttribute('id'))
    print(node.getAttribute('name'))
    print(node.getAttribute('author'))
    print(node.getAttribute('price'))
```

```python
# 将Python的字典转换为XML
from xml.etree.ElementTree import Element

def dict_to_xml(tag, d): 
    elem = Element(tag) # 创建一个根节点
    for key, val in d.items(): 
        child = Element(key) 
        child.text = str(val)
        elem.append(child)
    return elem
```

```python
# 同关系型数据库交互
# Python中表达行数据的标准方式是元组序列，比如：
data = [
    ('Alice', 25, 'F'),
    ('Bob', 30, 'M'),
    ('Charlie', 35, 'M'),
    ('David', 40, 'F'),
    ('Emma', 45, 'F'),
]

# 和不同数据库交互的接口是非常类似的，以MySQL为例，使用PyMySQL库，可以这样写：
import pymysql

# 连接到数据库
conn = pymysql.connect(host='localhost', user='root', password='password', db='mydatabase')

# 创建游标对象
cursor = conn.cursor()

# 执行SQL语句
sql = "INSERT INTO users (name, age, gender) VALUES (%s, %s,%s)"
cursor.executemany(sql, data)

# 提交事务
conn.commit()

# 查询操作
sql = "SELECT * FROM users"
cursor.execute(sql)
results = cursor.fetchall()
for row in results:
    print(row)

# 关闭游标和连接
cursor.close()
conn.close()
```
