# 迭代器和生成器
迭代器和生成器是Python中用于处理大量数据的重要概念。
迭代器是一种对象，使用iter()方法创建，并通过next()方法获取序列的下一个元素。当序列没有更多的元素时，next()方法会引发StopIteration异常。
生成器是一个特殊的迭代器，它使用yield关键字来定义。当调用生成器的next()方法时，生成器会返回一个值，然后暂停执行，直到再次调用next()方法。

```python
# 迭代器示例
l = iter([1, 2, 3, 4, 5])
while True:
    try:
        print(next(l))
    except StopIteration:
        print("迭代结束")
        break
```

```python
# 如果我们定义了一个类，希望它能支持迭代，那么，Python提供了一种机制，让一个类可以作为迭代器，即实现__iter__()方法。
class Fibs:
    def __init__(self,list=[]):
        self.a = []+list
    def __iter__(self):
        return iter(self.a)

fibs = Fibs([1,2,3])
for i in fibs:
    print(i)
```

```python
# 生成器示例，结社我们要生成一个等差数列的生成器
def arithmetic_sequence_generator(start, stop, increment):
    x = start
    while x <= stop:
        yield x
        x += increment
        
for i in arithmetic_sequence_generator(1, 10, 2):
    print(i)
    
print([a for a in (arithmetic_sequence_generator(2, 10, 2))])
```

```python
# 需要同时遍历多个序列，可以使用zip函数。zip函数将多个序列组合为一个元组序列。
a = [1, 2, 3]
b = [4, 5, 6]
c = [7, 8, 9]

print([(x,y,z) for x,y,z in zip(a,b,c)])
    
    

```

```python
# 需要先后遍历多个序列，可以使用itertools.chain()函数
from itertools import chain
a = [1, 2, 3]
b = [4, 5, 6]
print( [i for i in chain(a, b)])
    
# 或者使用heapq模块中的.merge()函数
import heapq
a = [1, 2, 3]
b = [4, 5, 6]
c = [7, 8, 9]
print([ x for x in heapq.merge(a, b, c)])
    
```

```python
# enumerate()函数可同时遍历元素和索引
for i, value in enumerate(['A', 'B', 'C']):
    print(i, value)
```
