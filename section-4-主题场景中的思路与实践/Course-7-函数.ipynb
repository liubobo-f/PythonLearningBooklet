```python
# 一个可接受任意数量参数的函数
def sum_numbers(*args):
    """计算传入参数的和"""
    return sum(args)

print(sum_numbers(1, 2, 3, 4, 5))

# 其中，传入的args是一个元组，元组中的元素个数由调用函数时传入的参数决定。
# 参数中也可以使用**kwargs，表示一个字典，字典中的键值对由调用函数时传入的参数决定。
```

```python
# 在定义函数的时候，建议加入额外的描述信息，帮助使用函数的用户了解函数的功能
def calculate_area(width:float, height:float)->float:
    """计算矩形的面积
    参数：
    width (float): 矩形的宽度
    height (float): 矩形的高度
    返回值：
    float: 矩形的面积
    """
    area = width * height
    return area

```

```python
# 让函数返回多个值
def ints():
    return 1, 2, 3
a, b, c = ints()
print(a, b, c)

# 函数将返回一个元组
```

```python
# 定义带有默认参数的函数，只要在定义时给参数设置默认值即可
def greet(name, greeting="Hello"):
    print(f"{greeting}, {name}!")
```

```python
# 定义匿名函数，简化代码
add = lambda x, y: x + y
print(add(1, 2))

# lambda表达式非常常用的场景是sorted()函数的key参数，用于指定排序的规则
word = ['apple', 'banana', 'peach']
word.sort(key=lambda x: len(x))
print(word)
```
