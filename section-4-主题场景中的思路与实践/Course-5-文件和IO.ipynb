```python
# print()方法可以将文件打印到文件中
with open('test.txt', 'xt') as f: # xt可以在文件不存在的情况下创建文件
    print('hello world', file=f)
```

```python
# 二进制文件的读写需要在文本读写的基础上增加编码和解码的步骤

with open('test.txt', 'rb') as f:
    data = f.read()
    text = data.decode('utf-8')
    
with open('test.txt', 'wb') as f:
    data = text.encode('utf-8')
    f.write(data)
```

```python
# 读写.gz等压缩过的文件中的数据，可以使用gzip模块
import gzip
with gzip.open('data.txt.gz', 'rb') as f:
    content = f.read()
    print(content)

# 读写.bz2等压缩过的文件中的数据，可以使用bz2模块
import bz2
with bz2.open('data.txt.bz2', 'rb') as f:
    content = f.read()

with bz2.open('data.txt.bz2', 'wb') as f:
    f.write(content)

# 写入时，bz2.open函数提供了compresslevel参数，可以设置压缩等级，默认为9，数值更低的等级会更快，但压缩率会更低


```

```python
# 处理文件名、目录、绝对路径可以使用os.path模块。而且尽量避免直接使用字符串操作，因为os.path考虑了系统的差异。

import os
path = '/Users/liubobo/Data/data.csv'

print("文件名",os.path.basename(path))
print("路径",os.path.dirname(path))
```

```python
# 判断文件或目录是否存在
print(os.path.exists('test.txt'))

# 获取文件的元数据
print(os.path.getctime('test.txt')) # 创建时间
print(os.path.getsize('test.txt')) # 文件大小
```

```python
# 获取目录中的文件列表
files = os.listdir(r"c:")
print(files)

# 如果希望筛选结果，可以使用推导式
files = [f for f in os.listdir(r"c:") if f.endswith(".Bin")]
print(files)
```
