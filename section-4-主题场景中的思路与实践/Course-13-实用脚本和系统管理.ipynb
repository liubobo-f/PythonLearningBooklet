```python
# 对文件和目录的操作通常可以使用shell完成，Python提供了shutil模块来实现。
import shutil
shutil.copy(src, dst)  # 复制文件或目录
shutil.copytree(src, dst)  # 复制目录
shutil.move(src, dst)  # 移动文件或目录
shutil.rmtree(path)  # 删除目录
```

```python
# 创建或解压.tar, .tgz, .zip文件，可以使用shutil的make_archive()和unpack_archive()函数
import shutil
# 创建tar文件
shutil.make_archive('my_archive', 'tar', 'my_directory')
# 解压tar文件
shutil.unpack_archive('my_archive.tar', 'extracted_directory')
# 创建zip文件
shutil.make_archive('my_archive', 'zip', 'my_directory')
# 解压zip文件
shutil.unpack_archive()
```

```python
# 查找文件可以使用 os.walk() 函数
# os.walk() 函数返回一个三元组，包含当前目录的路径、当前目录下的子目录和当前目录下的文件
import os
for dirpath, dirnames, filenames in os.walk('.'):
    if "fileToFind" in filenames:
        print(os.path.join(dirpath, "fileToFind"))
```

```python
# 对.ini格式配置文件的读写
# 假设我们有下面的配置文件：
# [section1]
# key1 = value1
# key2 = value2
# [section2]
# key3 = value3
# key4 = value4
#
# 我们可以使用 configparser 模块读取这个配置文件，并获取其中的值：
import configparser

config = configparser.ConfigParser()
config.read('config.ini')

print(config['section1']['key1'])  # 输出：value1

# 我们也可以使用 configparser 模块写入配置文件：
config.set('section1', 'key1', 'new value1')
with open('config.ini', 'w') as configfile:
    config.write(configfile)
    
# 在解析配置文件具体值的时候，建议使用 getint()、getfloat()、getboolean() 等方法
# 这些方法会自动将值转换为整数、浮点数、布尔值等类型，从而避免了类型转换的错误
# 比如getboolean()方法，如果配置文件中的值是true、false、yes、no、on、off，那么这些值都会被转换为True和False
```

```python
# 给脚本添加日志记录，可以使用 logging 模块
import logging
logging.basicConfig(
    level=logging.INFO
    , filename="log.txt"
    , filemode="w"
    , format="%(asctime)s - %(levelname)s - %(message)s"
)
logging.info("Hello, World!")
```

logging模块中，.critical()、.error()、.warning()、.info()、.debug()方法分别对应5个日志级别，
- 从高到低依次为：CRITICAL > ERROR > WARNING > INFO > DEBUG。
- 默认情况下，logging模块只输出INFO及以上级别的日志，即不会输出DEBUG级别的日志。
- 如果需要输出DEBUG级别的日志，需要设置日志级别为DEBUG，即logging.basicConfig(level=logging.DEBUG)。
- 如果需要输出所有级别的日志，可以设置日志级别为NOTSET，即logging.basicConfig

```python
# 如果想用配置文件配置log，可以改成下面的形式
import logging

logging.config.fileConfig('logging.conf')

# logging.conf文件内容如下，修改配置可以直接编辑logging.conf文件：
# [loggers]
# keys=root
#
# [handlers]
# keys=consoleHandler
#
# [formatters]
# keys=simpleFormatter
#
# [logger_root]
# level=DEBUG
# handlers=consoleHandler
#
# [handler_consoleHandler]
# class=StreamHandler
# level=DEBUG
# formatter=simpleFormatter
```
