# 网络和web编程

```python
# 发送get, post请求可以使用requests库
# 通常的步骤如下
import requests
parms_dic = {
    'name': 'zhangsan',
    'age': 18
}
headers = {
    'User-Agent': ''
}

res = requests.get('', params=parms_dic, headers=headers)
text = res.text
# res.content 获取二进制数据，res.json 获取json数据
```
