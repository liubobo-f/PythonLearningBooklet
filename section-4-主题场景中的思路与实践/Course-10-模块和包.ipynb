## 可以通过pyvenv命令创建虚拟环境
创建虚拟环境
python -m venv venv
激活虚拟环境
source venv/bin/activate
退出虚拟环境
deactivate
删除虚拟环境
rm -rf venv
创建虚拟环境，指定python版本
python3 -m venv --python=python3.6 venv
创建虚拟环境，指定虚拟环境名称
python3 -m venv --system-site-packages venv
