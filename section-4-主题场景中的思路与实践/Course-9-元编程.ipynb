```python
# 给函数加上包装层，实现函数的装饰
# 下面的实力展示了给函数计时的装饰器
import time
from functools import wraps
def time_it(func):
    @ wraps(func)
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print(func.__name__, end-start)
        return result
    return wrapper

@time_it
def countdown(n):
    while n > 0:
        n -= 1
countdown(5000000)

# 其中对functools.wraps的调用，是为了保留函数的元信息，比如函数名，函数的文档字符串等
```
