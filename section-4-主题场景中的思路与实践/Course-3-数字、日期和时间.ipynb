## 对数智做格式化输出，更一般的方式是使用format()函数
基本语法如下：
format(value, format_spec)

value：要格式化的值
format_spec：格式化规范，包括类型、精度、宽度等

format_spec的语法规则：
<type>[<width>][.<precision>][<type_modifier>]

type：格式化类型，包括：

d：整数
f：浮点数
e：科学计数法
g：科学计数法，如果小数点后没有数字，则使用f格式
s：字符串

width：宽度，表示输出字符串的最小长度，不足时前面补空格

```python
print(format(123, "08d"))
print(format(123.456, "08.2f"))
print(format("hello", ">8"))
```

## 随机相关操作可以使用random库

```python
# 随机相关操作可以使用random库
import random

# 从序列中挑选一个元素
print(random.choice([1, 2, 3, 4, 5]))
# 从序列中随机选择几个元素
print(random.sample([1, 2, 3, 4, 5], 3))
# 打乱序列
print(random.shuffle([1, 2, 3, 4, 5]))
```

```python
# 随机生成一个0-1的浮点数
print(random.random())
# 随机生成一个整数
print(random.randint(1, 10))
```

## 时间换算，可以使用datetime模块中的timedelta类

```python
# 时间换算，可以使用datetime模块中的timedelta类
from datetime import timedelta
a = timedelta(days=2, hours=6)
b = timedelta(hours=4.5)
c = a + b
print(c.days)
print(c.seconds)
print(c.total_seconds())
print(c.total_seconds() / 3600)
```

```python
# datetime中的timedelta类可以与datetime类、date类进行运算
from datetime import timedelta, datetime
now = datetime.now()
print(now + timedelta(days=3, hours=4))
```

```python
# 生成本月所有日期的列表
from datetime import date, timedelta
now = date.today()
start_date = date(now.year, now.month, 1)
end_date = date(now.year, now.month + 1, 1) - timedelta(days=1)
print([start_date + timedelta(days=x) for x in range((end_date - start_date).days + 1)])
```
