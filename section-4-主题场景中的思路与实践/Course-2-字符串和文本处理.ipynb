## 检查字符串开头和结尾
使用字符串的 startswith() 和 endswith() 方法

```python
# 检查字符串开头和结尾
txt = "A boy is playing"
print("是否以A开头",txt.startswith("A"))
print("是否以play结尾",txt.endswith("play"))

# 多选项检查
print("开头是否是A,B,C中的一个",any(txt.startswith(x) for x in ["A", "B", "C"]))
print("结尾是否是ed,ing中的一个",txt.endswith(("ed", "ing"))) # .endswith()接受元组
```

## 去掉收尾的字符

```python
# 去掉收尾字符的演示，strip()方法
str = "   hello world   "
print("!",str.strip(),"!",sep="")
print("!",str.lstrip(),"!",sep="")
print("!",str.rstrip(),"!",sep="")


str1 = "!!!!hello world"
print(str1.lstrip("!"))
```

```python
# 对齐字符串
str = "hello world"
print(str.ljust(20, "*"))
print(str.rjust(20))
```

```python
# 更一般的方法是使用format函数，可用在字符串以外的值上
print(format(str,">20"))
print(format(1.25,">20"))
```

```python
# 字符串的连接，使用join()
strlist = ['a', 'b', 'c']
print(','.join(strlist))
```
