# 条件语句
有时我们需要根据某些条件来执行不同的代码，这时我们就可以使用条件语句。
是否满足条件通常使用布尔值来表示，True表示满足条件，False表示不满足条件。判断大小是最常见的条件，Python提供了>、<、>=、<=、==、!=等操作符来判断大小关系。

特别注意：判断相等使用==，而不是=。在PYTHON中=仅是赋值操作符。

```python
print(2>=3)
print(2!=3)
```

PYTHON语句通常会顺序执行，当在特定条件下执行特定代码时，可以使用最简单的条件语句if ... else ...

举个例子，我们希望通过一个函数来判断某次考试成绩是否通过，如果成绩大于60分，则通过，否则不通过。

```python
def exam_status(score):
    if score >= 60:
        return '通过'
    else:
        return '不通过'
    
print(exam_status(80))
print(exam_status(50))
```

在上面的代码中，调用exam_status()函数会依此将80,50传入score，if语句会判断score是否大于等于60，如果是，继续执行return '通过'，否则执行else后的return '不通过'。（二选一）
另外，PYTHON使用缩进来区分代码块，缩进为4个空格，或者一个Tab键。if ... else ...是def下面的代码块，相对于def缩进4空格。return'通过”和return“不通过”是if ... else ...的代码块，相对于if缩进4空格。错误的缩进会导致程序运行逻辑混乱甚至报错。

同样的逻辑可以有不同的写法，比如下面代码块中的exam_status1()函数，可以实现与exam_status()函数相同的功能。注意：此时return的缩进与IF,ELSE相同，无论if ... else ...的代码块执行哪个分支，return的代码块都会执行。

```python
def exam_status(score):
    if score >= 60:
        output = '通过'
    else:
        output = '不通过'
    return output # 与if ... else ... 语句缩进相同，如果return缩进到else下一级，在score>=60时，exam_status()函数不返回任何值，即返回None。
    
print(exam_status(80))
print(exam_status(50))
```

if ... else ...语句还可改写成if ... elif ... else ...语句，elif是else if的缩写。

比如我们希望exam_status()函数接收到大于等于80分的成绩时返回"优秀"，大于等于60分小于80分的成绩时返回"良好"，小于60分的成绩时返回"一般"，那么代码如下：

```python
def exam_status(score):
    if score >= 80:
        return "优秀"
    elif score >= 60:
        return "良好"
    else:
        return "一般"

print(exam_status(90))
```

上面代码的执行逻辑：
1. 首先执行print()
2. 调用exam_status函数时，首先判断if语句中的score>=80，如果结果为True，执行if后面的代码，返回"优秀"
3. 如果if结果为False，判断elif中score>=60,执行elif后面的代码，返回"良好"
4. 如果elif结果为False，执行else后面的代码，返回"一般"。

对于if ... elif ... else语句
1. elif是可选的，可以没有，如果有多个elif，Python会从上到下依次判断条件，直到找到一个为True的分支，执行对应的代码块，并跳过其他elif和else。
2. else是可选的，可以没有。

```python
# 更多elif的场景示例：
def exam_status(score):
    if score>=80:
        return "优秀"
    elif score>=60:
        return "良好"
    elif score>=40:
        return "一般"
    else:
        return "不及格"
```
