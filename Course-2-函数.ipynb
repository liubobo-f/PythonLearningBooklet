# 函数
Notebook中的代码块有非常好的结构性，可以很方便的进行代码块的移动、复制、粘贴等操作。
函数就是一种特殊的代码块，你可以把多行代码封装成一个函数，然后重复使用。

```python
# 函数的创建方法
def add3_and_print(input_var):
    result = input_var + 3
    print(result)
    return result
```

这个函数可以接收一个数字，加3，打印，再返回加3后的结果。
其中开头的def定义了函数名，括号里是函数的参数，冒号后面是函数体。
函数体要缩进，包含了调用函数时的操作。
最后一行retuan 是函数运行后，返回的计算结果。

定义部分并不会执行函数，只有调用函数时才会执行。

```python
# 调用刚刚的函数
a = add3_and_print(1)
print("打印结果",a)
```

结果中第一行是函数内部打印的result变量，第二行是函数外部print函数的结果。
每次调用函数，都会使用已经定义好的代码块进行运算。
另外，return并不是函数必须的，如果函数没有return，那么函数的返回值就是None，函数中的代码块会全部执行，但是不会返回任何值。

```python
def showPI():
    print("PI = ", 3.1415926)

showPI()
```

举个例子，如果收入sallary需要交3%的税，我们可以设计一个sallary_after_tax函数，计算税后收入。

```python
def sallary_after_tax(sallary):
    return sallary*0.97

print(sallary_after_tax(10000))
```

函数可以没有参数，也可以有1个以上的参数。

```python
# 这是一个计算体积的函数
def volume(length, width, height):
    return length * width * height

print(volume(2, 3, 4))
```

```python
# 这是一个没有参数的函数
def greeting():
    print("Hello World!")

greeting()
```
