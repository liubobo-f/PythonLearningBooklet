# 类的特殊用法

# 类的魔术方法(Magic Methods)
 以双下划线__开头和结尾的方法称为魔术方法，它们在特定情况下会被自动调用

#  常见的魔术方法可以分为以下几类:
# 
#  1. 基本方法:
#     __init__(self,...): 构造方法,创建实例时调用
#     __str__(self): 返回对象的字符串表示,print()时调用
#     __repr__(self): 返回对象的"官方"字符串表示
#     __del__(self): 析构方法,对象被删除时调用
# 
#  2. 容器类方法:
#     __len__(self): 返回对象的长度,len()时调用
#     __getitem__(self,key): 实现对象的索引访问,obj[key]时调用
#     __setitem__(self,key,value): 实现对象的索引赋值,obj[key]=value时调用
#     __delitem__(self,key): 实现对象的索引删除,del obj[key]时调用
#     __contains__(self,item): 实现成员测试运算符in
# 
#  3. 迭代器方法:
#     __iter__(self): 返回迭代器对象,for循环时调用
#     __next__(self): 返回下一个值,迭代时调用
# 
#  4. 可调用对象方法:
#     __call__(self,...): 使实例能够像函数一样被调用
# 
#  5. 比较运算符方法:
#     __eq__(self,other): 实现==运算符
#     __ne__(self,other): 实现!=运算符
#     __lt__(self,other): 实现<运算符
#     __le__(self,other): 实现<=运算符
#     __gt__(self,other): 实现>运算符
#     __ge__(self,other): 实现>=运算符
# 
#  6. 算术运算符方法:
#     __add__(self,other): 实现+运算符
#     __sub__(self,other): 实现-运算符
#     __mul__(self,other): 实现*运算符
#     __truediv__(self,other): 实现/运算符
#     __floordiv__(self,other): 实现//运算符
#     __mod__(self,other): 实现%运算符
#     __pow__(self,other): 实现**运算符
#     __and__(self,other): 实现&运算符
#     __or__(self,other): 实现|运算符
#     __xor__(self,other): 实现^运算符
#     __lshift__(self,other): 实现<<运算符
#     __rshift__(self,other): 实现>>运算符
# 
#  7. 反向算术运算符方法:
#     __radd__(self,other): 实现反向+运算符
#     __rsub__(self,other): 实现反向-运算符
#     __rmul__(self,other): 实现反向*运算符
#     __rtruediv__(self,other): 实现反向/运算符
#     __rfloordiv__(self,other): 实现反向//运算符
#     __rmod__(self,other): 实现反向%运算符
#     __rpow__(self,other): 实现反向**运算符
# 
#  8. 类型转换方法:
#     __int__(self): 转换为int类型
#     __float__(self): 转换为float类型
#     __bool__(self): 转换为bool类型
#     __complex__(self): 转换为complex类型
# 
#  9. 上下文管理方法:
#     __enter__(self): 进入上下文管理器时调用
#     __exit__(self,exc_type,exc_value,traceback): 退出上下文管理器时调用
