# 环境和依赖
可以使用python自带的pyvenv 创建虚拟环境，也可以使用anaconda创建虚拟环境。

## pyvenv
- 创建虚拟环境
pyvenv venv

- 激活虚拟环境
source venv/bin/activate

- 退出虚拟环境
deactivate

- 删除虚拟环境
rm -rf venv

pip freeze 命令可以列出当前虚拟环境中安装的所有包，效果是生成requirements.txt文件

## pip
- 安装依赖
pip install -r requirements.txt

- 卸载依赖
pip uninstall <package>
