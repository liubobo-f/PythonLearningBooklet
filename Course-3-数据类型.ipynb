# 数据类型

每个变量对应一个数据类型。PYTHON内置许多不同的数据类型，如整数、浮点数、布尔值和字符串等，还可以自定义类型。type()函数可以查看变量的数据类型。

```python
# 整数类型int
var = 5
print(var,type(var))
```

```python
# 浮点数类型 float，即小数或分数
pi1 = 22/7
pi2 = 3.14
print(pi1,pi2)
print(type(pi1),type(pi2))
```

```python
# 布尔值类型 bool, 只有True和False两种值
var_bool = True
print(var_bool,type(var_bool))

var_bool2 = var<pi2 # 判断5<3.14，返回True或False
print(var_bool2,type(var_bool2))
```

```python
# 布尔值可以进行and,or,not等运算，结果也是布尔值。
print(var_bool and var_bool2) # True and False
print(var_bool or var_bool2) # True or False
print(not var_bool) # not True
```

```python
# 字符串类型
s = "Hello World"
print(s,type(s))
s1 = "73"
print(s1,type(s1))
```

```python
# 可以用len()函数来计算字符串长度
print(len(s),len(s1)) 
# 可以用+号来连接字符串
print(s+s1)
# 可以用*号来重复字符串
print(s*3)
```

变量之间可以相互转化，比如把一个字符串变量转化成数字变量，或者把一个数字变量转化成字符串变量。

```python
print(int("73"))
print(str(70+3)+" is my lucky number")
```

类型不匹配是一种非常常见的错误，Python在一些场景下会自动进行类型转换，但有时会报错。

```python
print(73+" is my lucky number")
```
