该系统可以实现以下功能：
1. 添加书籍：用户可以输入书籍的名称、作者、出版年份以及书籍类型，将书籍信息添加到系统中。
2. 查看所有书籍：系统可以显示当前所有已添加的书籍列表，并展示每本书籍的详细信息。
3. 删除书籍：用户可以通过书名删除指定的书籍。
4. 按书名查询书籍：用户可以通过书名查询具体书籍的详细信息。
5. 退出程序：用户可以选择退出程序，结束本次书籍管理操作。

参考代码如下：

功能定义部分：

```python
books = []  # 用来存储所有书籍信息

def add_book():
    title = input("请输入书名: ")
    author = input("请输入作者: ")
    year = input("请输入出版年份: ")
    book_type = input("请输入书籍类型（小说、历史、科技等）: ")
    
    # 创建书籍字典
    book = {
        'title': title,
        'author': author,
        'year': year,
        'type': book_type
    }
    
    # 添加到书籍列表
    books.append(book)
    print("书籍已成功添加！")

def list_books():
    if books:
        for book in books:
            print(f"书名: {book['title']}, 作者: {book['author']}, 年份: {book['year']}, 类型: {book['type']}")
    else:
        print("暂无书籍！")

def delete_book():
    title = input("请输入要删除的书名: ")
    for book in books:
        if book['title'] == title:
            books.remove(book)
            print("书籍已成功删除！")
            break


def query_book():
    title = input("请输入书名查询: ")
    found = False
    for book in books:
        if book['title'] == title:
            print(f"书名: {book['title']}, 作者: {book['author']}, 年份: {book['year']}, 类型: {book['type']}")
            found = True
            break
    if not found:
        print("未找到该书籍！")
```

主程序部分：

```python

print("\n欢迎使用书籍管理系统！请选择操作：")
print("1. 添加书籍")
print("2. 查看所有书籍")
print("3. 删除书籍")
print("4. 按书名查询书籍")
print("5. 退出程序")

while True:
    choice = input("请输入您的选择（1-5）: ")
    
    if choice == "1":
        add_book()
    elif choice == "2":
        list_books()
    elif choice == "3":
        delete_book()
    elif choice == "4":
        query_book()
    elif choice == "5":
        print("感谢使用书籍管理系统！")
        break
    else:
        print("无效的选择，请重新输入。")
        
    print("-"*20)
```

虽然我们的书籍管理系统目前能够实现基本的增、删、查操作，但它仍然可以通过以下方式进行扩展和优化：

2. `输入验证`：当前程序对用户输入没有进行严格的验证。例如，出版年份应该是数字，书名和作者不能为空等。你可以增加数据验证功能，提升程序的健壮性。
