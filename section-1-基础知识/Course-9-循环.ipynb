# 循环

循环是程序中非常常见的一种控制结构，用来重复执行某些代码块。
Python 中有两种常见的循环结构：`for` 循环和 `while` 循环。

## for 循环
`for` 循环用于遍历序列（如列表、元组、字典、集合、字符串等）中的每一个元素。遍历可以理解为一个循环，循环中的代码会依次执行，直到循环结束。

for item in sequence:
    # 循环体

其中，`item` 是一个变量，用于存储序列中的每个元素。`sequence` 是一个序列，可以是列表、元组、字典、集合、字符串等。
for循环可以将循环体中的代码块执行多次。

以下是一个简单的例子，使用 `for` 循环遍历一个列表：

```python
for fruit in ['apple', 'banana', 'orange']:
    print(fruit)
```

如果列表中是一些文件，那就遍历了这些文件，每次循环操作列表中的一个文件。

当想循环特定的次数时，可以配合range()函数。
`range()` 函数返回一个序列，可以用于生成一定范围的整数。默认从 0 开始，步长为 1，直到指定的结束值（不包括结束值）。
注意：range()函数返回的是一个可迭代对象，而不是一个列表（不可以直接打印）。要得到一个列表，需要使用list()函数将其转换为列表。

```python
for i in range(5):  # 输出从 0 到 4
    print(i)
```

for循环可以遍历列表、元组、集合、字典等数据类型。

```python
person = {"name": "Alice", "age": 25, "city": "New York"}
person.items()
```

```python
# 遍历字典中的键值对
person = {"name": "Alice", "age": 25, "city": "New York"}
for key, value in person.items():
    print(f"键：{key}，值：{value}")
```

```python
# For循环还经常用来快速创建列表或其他数据结构
squares = [x**2 for x in range(10)]
string_lengths = [len(s) for s in ['apple', 'banana', 'cherry']]
tuplelist = [ (x, len(x)) for x in ['apple', 'banana', 'cherry']]

print("平方数列表：", squares)
print("字符串长度列表：", string_lengths)
print("元组列表：", tuplelist)
```

如果我们有一个提取文件大小的函数，那么我们可以使用一个循环来遍历所有文件并计算它们的大小。比如：
[size(file) for file in files]

这里也可以使用List的方法： files.map(size)

两种方法均比使用循环更简洁。循环的写法：
for file in files:
    print(size(file))

## while循环
`while` 循环用于当某个条件为 True 时重复执行代码块。它在条件为 False 时停止执行。

```python
# 一个简单的 while 循环
count = 0
while count < 5:
    print(count)
    count += 1  # 等价于 count = count + 1，这个循环会执行到count = 5时结束
```

如果条件一直未True，循环将一直执行下去，使用`break`可以退出循环。

```python
while True:
    user_input = input("请输入 'q' 退出循环：")
    if user_input == 'q':
        break  # 当用户输入 'q' 时，退出循环
    print(f"你输入了: {user_input}")
```

使用 `continue` 跳过当前循环的余下部分，进入下一次循环。

```python
for i in range(5):
    if i == 3:
        continue  # 跳过 i 等于 3 时的输出
    print(i)  # 输出 0, 1, 2, 4
```

循环可以有一个 `else` 子句，它在循环正常结束时执行（即没有遇到 `break`）。

```python
for i in range(3):
    print(i)
else:
    print("for 循环正常结束。")  # 这个语句会在循环结束后执行。
```

循环可以用来求和

```python
# 计算 1 到 10 的和
total = 0
for i in range(1, 11):
    total += i
print("1 到 10 的和是:",total)
```

当然，上述的求和也可以写成下面的形式

```python
s = sum(list(range(11)))
print("1 到 10 的和是:",s)
```

```python
sum([x for x in range(11)])  # 使用列表推导式计算和
```

# 总结：
- `for` 循环用于遍历序列（如列表、元组、字典、字符串等）中的每个元素。
- `while` 循环用于在条件为 True 时重复执行某些代码块。
- `break` 用于提前结束循环，`continue` 用于跳过当前循环。
- `else` 子句在循环正常结束时执行。
- 循环常用于计算、判断和生成特定数据结构等。
