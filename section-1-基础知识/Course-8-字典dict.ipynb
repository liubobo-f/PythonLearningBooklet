# 字典 dict
字典是一个无序的键值对集合。每个键（key）都对应一个值（value）。
字典是可变的，可以修改键值对。字典中的键是唯一的，但值可以重复。

```python
# 字典使用大括号 {} 来表示，键值对之间用冒号 ":" 隔开，多个键值对之间用逗号 "," 隔开。
dict1 = {"name": "Alice", "age": 25, "city": "New York"}
dict2 = {"fruit": "apple", "color": "red", "quantity": 5}

# 打印字典
print("dict1:", dict1)
print("dict2:", dict2)
```

```python
# 通过键来访问字典中的值。如果键不存在，会引发 KeyError。
print("dict1 中 'name' 对应的值:", dict1["name"])  
print("dict2 中 'fruit' 对应的值:", dict2["fruit"]) 

# 如果键不存在，可以使用 get() 方法避免 KeyError。get() 返回 None 或指定的默认值。
print("dict1 中 'gender' 的值:", dict1.get("gender", "Unknown")) 
```

```python
# 可以通过键来修改字典中的值，如果键不存在，则会新增该键值对。
dict1["age"] = 26
print("修改 'age' 后的 dict1:", dict1)

# 添加新的键值对
dict1["email"] = "alice@example.com"
print("添加 'email' 后的 dict1:", dict1)

# 使用 del 删除指定键的键值对。
del dict1["city"]
print("删除 'city' 后的 dict1:", dict1)
```

```python
# 使用 len() 函数获取字典中键值对的个数。
print("dict1 的长度（键值对数量）：", len(dict1))
```

```python
# 使用 keys() 方法获取字典的所有键。
print("dict1 的所有键：", dict1.keys())

# 使用 values() 方法获取字典的所有值。
print("dict1 的所有值：", dict1.values())

# 使用 items() 方法获取字典的所有键值对。
print("dict1 的所有键值对：", dict1.items())
```

```python
# 可以通过 update() 方法将另一个字典的键值对添加到当前字典中。
dict1.update(dict2)
print("dict1 和 dict2 合并后的结果：", dict1)
```

```python
# 字典可以包含其他字典，这种情况叫做嵌套字典。
nested_dict = {
    "person1": {"name": "Alice", "age": 25},
    "person2": {"name": "Bob", "age": 30}
}
print("嵌套字典：", nested_dict)

# 访问嵌套字典的值
print("person1 的年龄：", nested_dict["person1"]["age"]) 
```

```python
# clear() 删除字典中的所有键值对。
dict1.clear()
print("清空后的 dict1：", dict1)
```

# 总结
- 字典是一个无序的键值对集合，键是唯一的，值可以重复。
- 字典支持修改、添加、删除键值对。
- 字典有丰富的内置方法，可以获取键、值、键值对，进行遍历、合并等操作。
- 字典也可以嵌套使用，可以创建更复杂的数据结构。
