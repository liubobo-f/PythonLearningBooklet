# 集合 set
集合是一个无序的、不可重复的元素集。集合中的元素是唯一的，并且没有索引。
集合常用于需要去重或测试集合间关系（如交集、并集等）时。

```python
# 集合用大括号 {} 来表示，元素之间用逗号隔开。集合中的元素是唯一的，不允许重复。
set1 = {1, 2, 3, 4}
set2 = {"apple", "banana", "cherry"}
set3 = {1, "apple", 3.14, (5, 6)}  # 集合可以包含不同类型的元素，但不可包含可变元素，如列表
```

```python
# 打印集合
print("set1:", set1)
print("set2:", set2)
print("set3:", set3)
```

```python
# 使用 add() 方法可以向集合中添加一个元素。
set1.add(5)
print("添加元素 5 后的 set1:", set1)
```

```python
# 使用 remove() 方法删除一个元素，如果元素不存在，会引发 KeyError。
# 使用 discard() 方法删除元素，如果元素不存在，不会引发错误。
set2.remove("banana")
print("删除 'banana' 后的 set2:", set2)

set2.discard("orange")  # 不存在的元素
print("删除不存在的 'orange' 后的 set2:", set2)
```

```python
# 使用 len() 函数获取集合的元素个数。
print("set1 的长度：", len(set1)) 
```

```python
# 使用 in 关键字检查元素是否在集合中。
print("3 是否在 set1 中？", 3 in set1)  
print("'apple' 是否在 set2 中？", "apple" in set2)  
```

集合的运算

```python
# 交集：返回两个集合中都存在的元素
setA = {1, 2, 3, 4}
setB = {3, 4, 5, 6}
print("setA 和 setB 的交集：", setA & setB) 

# 并集：返回两个集合中所有的元素，去重
print("setA 和 setB 的并集：", setA | setB)

# 差集：返回在 setA 中但不在 setB 中的元素
print("setA 和 setB 的差集：", setA - setB) 

# 对称差集：返回两个集合中不重复的元素
print("setA 和 setB 的对称差集：", setA ^ setB)
```

```python
print(setA.intersection(setB))
```

上述四种运算也可以使用集合的方法，比如交集使用 intersection() 方法，并集使用 union() 方法，差集使用 difference() 方法，对称差集使用 symmetric_difference() 方法。

```python
# 集合的不可变版本：frozenset
# frozenset 是不可变的集合类型，一旦创建不能修改。
frozen_set = frozenset([1, 2, 3, 4])
print("frozenset:", frozen_set)
```

## 总结：
- 集合是无序的，不允许重复元素。
- 集合支持交集、并集、差集等集合运算。
- 集合的成员是不可重复的，并且不能有索引。
- 可以使用 frozenset 创建不可变的集合。

List, Tuple, Set的区别是：
List 是一个有序的集合，可以随时添加和删除其中的元素。
Tuple 是一个有序的集合， tuple一旦被创建，便不可更改。
Set 是一个无序的集合，集合中不能有重复的元素。

在选择使用哪种类型时，需要考虑：
1. 元素的顺序是否很重要？
2. 元素是否重复？
3. 元素的顺序是否需要保持不变？
4. 元素是否需要去重？
5. 元素是否需要频繁添加和删除？
