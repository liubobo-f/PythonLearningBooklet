# 列表list
当我们希望用一个变量来保存多个值时，PYTHON提供了数种数据类型，其中最常用的就是列表list。

假设我们有如下字符串：

```python
students = "Alice,Bob,Charlie,David,Emma,Frank,Gregory,Hannah,Isabella,Jack"
print(students,type(students))
```

类似的数据可以存储在字符串中，也可以以使用列表来存储。用[]包裹多个数据，用逗号隔开，每个数据成为列表的元素，每个元素可以是任意类型。

```python
students_list = ['Alice', 'Bob', 'Charlie', 'David', 'Emma', 'Frank', 'Gregory', 'Hannah', 'Isabella', 'Jack']
print(students_list,type(students_list))
```

虽然保存了相同的信息量，但列表类型提供了丰富的方法，使得对数据进行操作更加方便。

```python
# 计算列表中元素数
print("列表中有", len(students_list), "个元素")

# 在列表后使用方括号，可索引到列表中的具体元素
print("第一个元素是", students_list[0])
print("第二个元素是", students_list[1])
print("最后一个元素是", students_list[-1]) # 列表索引从0开始，-1表示最后一个元素，-2表示倒数第二个元素，以此类推。

# 甚至可以选取列表的一个片段，用冒号分隔开
print("第二个元素到第四个元素是", students_list[1:4]) # 列表索引从0开始，[]中最后一个数字表示的索引是不包含在内的，即左闭右开区间。
print("最后一个元素到倒数第二个元素是", students_list[-2:]) # 末尾可以省略，表示到列表末尾
print("第一个元素到倒数第二个元素是", students_list[:-1]) # 开始可以省略，表示从列表开头

```

```python
# 列表可以进行修改，比如修改列表中的某个元素
students_list[0] = "刘波波"
print("修改后的第一个元素是", students_list[0])

# 列表可以进行追加，比如在列表末尾追加一个元素
students_list.append("刘波波")
print("追加后的列表是", students_list)

# 列表可以进行反转，比如对列表进行反转
students_list.reverse()
print("反转后的列表是", students_list)

# 列表可以进行删除，比如删除列表中的某个元素
students_list.remove("刘波波")
print("删除后的列表是", students_list)
```

```python
# 如果一个列表中都是数字，那么你可以使用sum()，min()，max()函数来计算列表中所有数字的统计值
number_list= [1, 2, 3, 4, 5]
print("列表中所有数字的和是", sum(number_list))
print("列表中所有数字的最小值是", min(number_list))
print("列表中所有数字的最大值是", max(number_list))

# 列表可以进行排序，比如对列表进行降序排序
number_list.sort(reverse=True) # 不带参数是升序，reverse=True是降序
print("排序后的列表是", number_list)
```

将有明确分隔符的字符串转变成列表的方式如下：

```python
name_list = students.split(",") # 参数为分隔符
print(students)
print(name_list,type(name_list))
```
