# 元组 tuple
元组是 Python 中的一个数据类型，它是一个不可变的序列。与列表不同，元组一旦创建，就不能修改其内容。
元组通常用于存储不同类型的元素，比如可以包含字符串、整数、浮点数、甚至其他元组。

```python
# 元组用小括号 () 来表示，元素之间用逗号隔开。
tuple1 = (1, 2, 3, 4)
tuple2 = ("apple", "banana", "cherry")
tuple3 = (1, "apple", 3.14, (5, 6))
```

```python
# 打印元组
print("tuple1:", tuple1)
print("tuple2:", tuple2)
print("tuple3:", tuple3)
```

```python
# 可以通过索引来访问元组中的元素。索引从 0 开始。
print("tuple1 的第一个元素：", tuple1[0]) 
print("tuple2 的第二个元素：", tuple2[1])
```

```python
# 与列表一样，元组也支持切片操作。
print("tuple1 的前两个元素：", tuple1[:2])
```

```python
# 尝试修改元组中的元素会报错，因为元组是不可变的。
try:
    tuple1[0] = 10  # 这将导致错误
except TypeError as e:
    print("错误:", e)
```

```python
# 使用 len() 函数可以获取元组的长度（元素个数）。
print("tuple1 的长度：", len(tuple1)) 
```

```python
# 可以通过 + 运算符将多个元组合并成一个新的元组。
tuple4 = tuple1 + tuple2
print("tuple1 和 tuple2 连接后的结果：", tuple4)
```

```python
# 使用 * 运算符可以将元组重复指定次数。
tuple5 = tuple2 * 2
print("tuple2 重复两次后的结果：", tuple5)
```

```python
# 使用 in 关键字检查元素是否在元组中。
print("2 是否在 tuple1 中？", 2 in tuple1)
print("'apple' 是否在 tuple2 中？", "apple" in tuple2) 
```

```python
# 可以将元组的元素解包为多个变量。
a, b, c, d = tuple1
print("解包后的结果：", a, b, c, d)
```

```python
# 元组和列表可以互相转化。
list_from_tuple = list(tuple1)  
print("从 tuple1 转换为列表：", list_from_tuple)
```

# 总结：
- 元组是不可变的序列，一旦创建就不能更改。
- 元组可以包含多种类型的元素。
- 元组可以进行索引、切片、连接、重复等操作。
- 元组支持解包，可以将其元素分配给多个变量。
