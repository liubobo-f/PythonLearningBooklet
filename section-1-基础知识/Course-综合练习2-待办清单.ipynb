# 待办清单综合练习
欢迎来到本课程第一阶段的练习——制作一个简易待办清单软件。



一个基础版本的待办清单包括如下功能：
1. 添加待办事项
2. 删除待办事项
3. 标记待办事项为完成

实现思路：
1. 创建一个List，作为待办清单。
2. 创建一个函数，用于添加待办事项。
3. 创建一个函数，用于删除待办事项。
4. 创建一个函数，用于标记待办事项为完成。
5. 创建一个主函数，用于调用上述函数。
    1. 创建一个while循环，用于重复执行程序。
    2. 在循环中，打印菜单，提示用户选择操作。
    3. 根据用户选择，调用对应的函数。
    4. 循环结束后，打印退出信息。


参考代码如下：

功能定义部分：

```python
'''
实现思路：
1. 创建一个List，作为待办清单。
2. 创建一个函数，用于添加待办事项。
3. 创建一个函数，用于删除待办事项。
4. 创建一个函数，用于标记待办事项为完成。
5. 创建一个主函数，用于调用上述函数。
'''

todo_list = []
def add_todo():
    """添加待办事项"""
    item = input("请输入待办事项: ")    
    todo_list.append(item)
    print(f"已添加待办事项: {item}")
    
def delete_todo():
    """删除待办事项"""
    item = input("请输入要删除的待办事项: ")
    if item in todo_list:
        todo_list.remove(item)
        print(f"已删除待办事项: {item}")
    else:
        print(f"待办事项 '{item}' 不在清单中。")
        
def complete_todo():
    """标记待办事项为完成"""
    item = input("请输入已完成的待办事项: ")
    if item in todo_list:
        todo_list.remove(item)
        print(f"已标记待办事项 '{item}' 为完成。")
    else:
        print(f"待办事项 '{item}' 不在清单中。")
        
def show_todos():
    """显示待办清单"""
    if todo_list:
        print("当前待办清单:")
        for idx, item in enumerate(todo_list, start=1):
            print(f"{idx}. {item}")
    else:
        print("待办清单为空。")
              
```

```python
# 主程序

while True:
    print("\n待办事项管理系统")
    print("1. 添加待办事项")
    print("2. 删除待办事项")
    print("3. 标记待办事项为完成")
    print("4. 显示待办清单")
    print("5. 退出")
    
    choice = input("请选择操作 (1-5): ")
    
    if choice == '1':
        add_todo()
    elif choice == '2':
        delete_todo()
    elif choice == '3':
        complete_todo()
    elif choice == '4':
        show_todos()
    elif choice == '5':
        print("退出系统。")
        break
    else:
        print("无效的选择，请重新输入。")
        
        
```
