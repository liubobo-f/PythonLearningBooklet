# Print
我们从打印入手，使用print()并放入一个带引号的字符串，来打印这个字符串。
在notebook中，结果会显示在输出框下面，只要按Ctrl+Enter就可以运行。


```python
print("Hello, world")
```

在print中放入公式也是可行的，print()会计算出输出的公式的值，然后输出。
Python的基本运算符包括加（+）、减（-）、乘（*）、除（/）、取余（%）、取整除（//）、幂（**）等。

```python
print(2*3)
```

```python
# 使用括号可以控制公式的运算顺序。
print(2**(3+1))
```

你还可以写注释来解释代码的运行逻辑，让读者更好地理解代码的含义，并方便后续维护。代码数量越大，注释越重要。
如上个格子里一样，注释可以写在#后，不会执行。

print函数中可以用逗号间隔放入多个参数，会以空格间隔打印。

```python
print("我最喜欢的数字是", 73)
```

# 变量
变量可以把计算结果存储起来，下次用到的时候直接调用变量。创建变量的方法如下:

```python
my_first_var = 2*3
```

创建变量有个几个注意事项：
1. 变量名称不能以数字开头，只能以字母开头。
2. 变量名称不能以特殊符号开头。
3. 变量名称不能以空格开头。
4. 变量名称不能以特殊符号结尾。
5. 变量名称不能以空格结尾

变量的基本操作包括，创建、打印、修改等，后面我们还会学到更多的变量操作。
另外，变量名最好简短并清楚的描述它的作用。

```python
# 创建变量
length = 2
width = 3
area = length * width
# 修改变量
area = area * 2
# 打印变量
print(area)
```

```python
# 如果我们在之前的单元格中创建的变量，后面的单元格也可以访问到
# 再次打印area
print(area)
```
